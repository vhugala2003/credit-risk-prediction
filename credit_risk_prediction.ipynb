```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("credit_risk_dataset.csv")

# Display the first 5 rows
print("First 5 rows of the dataset:")
display(df.head())

# Dataset shape
print("Dataset Shape:")
print(df.shape)

# Dataset columns
print("Dataset Columns:")
print(df.columns.tolist())

# Dataset information
print("Dataset Information:")
df.info()

# Statistical summary
print("Statistical Summary:")
display(df.describe())

# Missing values
print("Missing Values:")
display(df.isnull().sum())

# Visualisation 1: Target Variable Distribution

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
sns.countplot(x='loan_status', data=df)

plt.title('Distribution of Loan Status')
plt.xlabel('Loan Status')
plt.ylabel('Number of Applicants')
plt.show()

# Interpretation:
# This graph shows the number of applicants in each loan status category.
# 0 represents applicants who did not default, while 1 represents applicants who defaulted.
# The graph helps us see whether the target variable is balanced or imbalanced.

# Visualisation 2: Correlation Heatmap

plt.figure(figsize=(10, 7))

# Select only numerical columns
numeric_df = df.select_dtypes(include='number')

# Create correlation matrix
correlation = numeric_df.corr()

# Draw heatmap
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap of Numerical Features')
plt.show()

# Interpretation:
# The heatmap shows the relationship between numerical variables.
# Values close to +1 show a strong positive relationship.
# Values close to -1 show a strong negative relationship.
# Values close to 0 show a weak or no linear relationship.

# Visualisation 3: Boxplots for Outlier Detection

numerical_features = [
    'person_age',
    'person_income',
    'person_emp_length',
    'loan_amnt',
    'loan_int_rate',
    'loan_percent_income',
    'cb_person_cred_hist_length'
]

plt.figure(figsize=(15, 10))

for i, column in enumerate(numerical_features, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=df[column])
    plt.title(column)

plt.tight_layout()
plt.show()

# Interpretation:
# The boxplots help identify possible outliers in the numerical features.
# Values shown as individual points outside the whiskers may be potential outliers.
# These outliers will be examined and treated during the preprocessing stage.

# Visualisation 4: Histograms of Numerical Features

selected_features = [
    'person_age',
    'person_income',
    'loan_amnt',
    'loan_int_rate',
    'loan_percent_income'
]

plt.figure(figsize=(15, 10))

for i, column in enumerate(selected_features, 1):
    plt.subplot(2, 3, i)
    sns.histplot(df[column], kde=True)
    plt.title(column)
    plt.xlabel(column)
    plt.ylabel('Frequency')

plt.tight_layout()
plt.show()

# Interpretation:
# The histograms show how the numerical features are distributed.
# They help us identify whether the data is normally distributed,
# skewed, or concentrated around certain values.
# These distributions will help us understand the data before preprocessing.

# Check missing values before preprocessing

print("Missing values before preprocessing:")
display(df.isnull().sum())

# Handle missing values

# Numerical columns
numerical_columns = df.select_dtypes(include='number').columns

for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())

# Categorical columns
categorical_columns = df.select_dtypes(include='object').columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

print("Missing values after preprocessing:")
display(df.isnull().sum())

# Interpretation:
# Numerical missing values are replaced with the median because
# the median is less affected by extreme values.
# Categorical missing values are replaced with the most frequent category.

# Check for duplicate records

print("Number of duplicate rows before removal:")
print(df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

print("\nNumber of duplicate rows after removal:")
print(df.duplicated().sum())

print("\nDataset shape after removing duplicates:")
print(df.shape)

# Interpretation:
# Duplicate records can cause the same applicant to be counted more than once.
# Removing duplicates helps prevent the model from learning from repeated records.

# Treat outliers using the IQR method

numerical_columns = df.select_dtypes(include='number').columns

# Exclude the target variable from outlier treatment
numerical_features = [
    column for column in numerical_columns
    if column != 'loan_status'
]

for column in numerical_features:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Cap values outside the boundaries
    df[column] = df[column].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Outlier treatment completed.")

# Interpretation:
# The IQR method identifies extreme values using the first and third quartiles.
# Values below the lower boundary or above the upper boundary are capped.
# The target variable (loan_status) is not changed because it is the variable
# we want the machine learning models to predict.

# Encode categorical variables using One-Hot Encoding

categorical_columns = df.select_dtypes(include='object').columns

print("Categorical columns before encoding:")
print(categorical_columns.tolist())

# Convert categorical variables into numerical variables
df = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)

print("\nDataset after encoding:")
display(df.head())

print("\nNew dataset shape:")
print(df.shape)

# Interpretation:
# Categorical variables have been converted into numerical columns
# using one-hot encoding.
# drop_first=True removes one category from each variable to avoid
# unnecessary redundancy between the encoded columns.

# Separate features and target variable

X = df.drop('loan_status', axis=1)
y = df['loan_status']

print("Features (X) shape:")
print(X.shape)

print("\nTarget (y) shape:")
print(y.shape)

print("\nTarget variable:")
display(y.head())

# Split the dataset into training and testing sets

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set shape:")
print(X_train.shape)

print("\nTesting set shape:")
print(X_test.shape)

print("\nTraining target shape:")
print(y_train.shape)

print("\nTesting target shape:")
print(y_test.shape)

# Interpretation:
# 80% of the data is used to train the machine-learning models.
# 20% is kept aside to evaluate the models on unseen data.
# stratify=y keeps the proportion of loan defaults similar in both sets.

# Feature Scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit the scaler only on the training data
X_train_scaled = scaler.fit_transform(X_train)

# Use the same scaler to transform the test data
X_test_scaled = scaler.transform(X_test)

print("Original training data shape:")
print(X_train.shape)

print("\nScaled training data shape:")
print(X_train_scaled.shape)

print("\nScaled testing data shape:")
print(X_test_scaled.shape)

# Interpretation:
# Feature scaling puts the features on a similar numerical scale.
# This is important for algorithms such as Logistic Regression and kNN,
# because features with very large values could otherwise have a greater influence.
# The scaler is fitted only on the training data to prevent data leakage.

# Part D: Principal Component Analysis (PCA)

from sklearn.decomposition import PCA
import numpy as np

# Apply PCA to the scaled training data
pca = PCA()

X_train_pca = pca.fit_transform(X_train_scaled)

# Transform the test data using the same PCA
X_test_pca = pca.transform(X_test_scaled)

print("Original number of features:")
print(X_train.shape[1])

print("\nNumber of PCA components:")
print(X_train_pca.shape[1])

print("\nExplained variance ratio:")
print(pca.explained_variance_ratio_)

# Cumulative Explained Variance

cumulative_variance = np.cumsum(pca.explained_variance_ratio_)

plt.figure(figsize=(10, 6))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker='o'
)

plt.axhline(
    y=0.90,
    linestyle='--',
    label='90% Variance'
)

plt.axhline(
    y=0.95,
    linestyle='--',
    label='95% Variance'
)

plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Cumulative Explained Variance using PCA')

plt.legend()
plt.grid()
plt.show()

# Interpretation:
# The graph shows how much total variance is retained as more
# principal components are added.
# The 90% and 95% lines help us identify the number of components
# needed to retain most of the information in the original dataset.

# Find the number of components needed for 90% and 95% variance

components_90 = np.argmax(cumulative_variance >= 0.90) + 1
components_95 = np.argmax(cumulative_variance >= 0.95) + 1

print("Components needed for at least 90% variance:", components_90)
print("Components needed for at least 95% variance:", components_95)

print("\nVariance explained by 90% selection:")
print(cumulative_variance[components_90 - 1])

print("\nVariance explained by 95% selection:")
print(cumulative_variance[components_95 - 1])

# Create PCA dataset using 95% variance

pca_95 = PCA(n_components=components_95)

X_train_pca_95 = pca_95.fit_transform(X_train_scaled)
X_test_pca_95 = pca_95.transform(X_test_scaled)

print("Original number of features:", X_train.shape[1])
print("PCA number of features:", X_train_pca_95.shape[1])

print("\nPCA training dataset shape:")
print(X_train_pca_95.shape)

print("\nPCA testing dataset shape:")
print(X_test_pca_95.shape)

print("\nTotal variance retained:")
print(pca_95.explained_variance_ratio_.sum())

# Interpretation:
# PCA has reduced the number of features while retaining at least 95%
# of the information contained in the original features.
# The PCA-transformed data will be used to train and compare
# machine-learning models.

# Part E: Machine Learning Models

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("Machine learning models and evaluation tools imported successfully.")

# Logistic Regression using Original Features

logistic_original = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model
logistic_original.fit(X_train_scaled, y_train)

# Make predictions
y_pred_logistic_original = logistic_original.predict(X_test_scaled)

# Calculate performance metrics
accuracy = accuracy_score(y_test, y_pred_logistic_original)
precision = precision_score(y_test, y_pred_logistic_original)
recall = recall_score(y_test, y_pred_logistic_original)
f1 = f1_score(y_test, y_pred_logistic_original)

print("Logistic Regression - Original Features")
print("---------------------------------------")
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_logistic_original))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_logistic_original))

# Part E: Machine Learning Models

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("Machine learning models and evaluation tools imported successfully.")

# Logistic Regression using Original Features

logistic_original = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model
logistic_original.fit(X_train_scaled, y_train)

# Make predictions
y_pred_logistic_original = logistic_original.predict(X_test_scaled)

# Calculate performance metrics
accuracy = accuracy_score(y_test, y_pred_logistic_original)
precision = precision_score(y_test, y_pred_logistic_original)
recall = recall_score(y_test, y_pred_logistic_original)
f1 = f1_score(y_test, y_pred_logistic_original)

print("Logistic Regression - Original Features")
print("---------------------------------------")
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_logistic_original))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_logistic_original))

# Logistic Regression using PCA Features

logistic_pca = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model using PCA features
logistic_pca.fit(X_train_pca_95, y_train)

# Make predictions
y_pred_logistic_pca = logistic_pca.predict(X_test_pca_95)

# Calculate performance metrics
accuracy = accuracy_score(y_test, y_pred_logistic_pca)
precision = precision_score(y_test, y_pred_logistic_pca)
recall = recall_score(y_test, y_pred_logistic_pca)
f1 = f1_score(y_test, y_pred_logistic_pca)

print("Logistic Regression - PCA Features")
print("-----------------------------------")
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_logistic_pca))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_logistic_pca))

# k-Nearest Neighbors (kNN) using Original Features

knn_original = KNeighborsClassifier(n_neighbors=5)

# Train the model
knn_original.fit(X_train_scaled, y_train)

# Make predictions
y_pred_knn_original = knn_original.predict(X_test_scaled)

# Calculate performance metrics
accuracy = accuracy_score(y_test, y_pred_knn_original)
precision = precision_score(y_test, y_pred_knn_original)
recall = recall_score(y_test, y_pred_knn_original)
f1 = f1_score(y_test, y_pred_knn_original)

print("kNN - Original Features")
print("-----------------------")
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_knn_original))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_knn_original))

# k-Nearest Neighbors (kNN) using PCA Features

knn_pca = KNeighborsClassifier(n_neighbors=5)

# Train the model
knn_pca.fit(X_train_pca_95, y_train)

# Make predictions
y_pred_knn_pca = knn_pca.predict(X_test_pca_95)

# Calculate performance metrics
accuracy = accuracy_score(y_test, y_pred_knn_pca)
precision = precision_score(y_test, y_pred_knn_pca)
recall = recall_score(y_test, y_pred_knn_pca)
f1 = f1_score(y_test, y_pred_knn_pca)

print("kNN - PCA Features")
print("------------------")
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_knn_pca))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_knn_pca))

# Decision Tree using Original Features

tree_original = DecisionTreeClassifier(
    random_state=42
)

# Train the model
tree_original.fit(X_train_scaled, y_train)

# Make predictions
y_pred_tree_original = tree_original.predict(X_test_scaled)

# Calculate performance metrics
accuracy = accuracy_score(y_test, y_pred_tree_original)
precision = precision_score(y_test, y_pred_tree_original)
recall = recall_score(y_test, y_pred_tree_original)
f1 = f1_score(y_test, y_pred_tree_original)

print("Decision Tree - Original Features")
print("---------------------------------")
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_tree_original))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_tree_original))

# Decision Tree using PCA Features

tree_pca = DecisionTreeClassifier(
    random_state=42
)

# Train the model
tree_pca.fit(X_train_pca_95, y_train)

# Make predictions
y_pred_tree_pca = tree_pca.predict(X_test_pca_95)

# Calculate performance metrics
accuracy = accuracy_score(y_test, y_pred_tree_pca)
precision = precision_score(y_test, y_pred_tree_pca)
recall = recall_score(y_test, y_pred_tree_pca)
f1 = f1_score(y_test, y_pred_tree_pca)

print("Decision Tree - PCA Features")
print("----------------------------")
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_tree_pca))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_tree_pca))

# Compare all six models

results = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Logistic Regression',
        'kNN',
        'kNN',
        'Decision Tree',
        'Decision Tree'
    ],
    'Features': [
        'Original',
        'PCA',
        'Original',
        'PCA',
        'Original',
        'PCA'
    ],
    'Accuracy': [
        accuracy_score(y_test, y_pred_logistic_original),
        accuracy_score(y_test, y_pred_logistic_pca),
        accuracy_score(y_test, y_pred_knn_original),
        accuracy_score(y_test, y_pred_knn_pca),
        accuracy_score(y_test, y_pred_tree_original),
        accuracy_score(y_test, y_pred_tree_pca)
    ],
    'Precision': [
        precision_score(y_test, y_pred_logistic_original),
        precision_score(y_test, y_pred_logistic_pca),
        precision_score(y_test, y_pred_knn_original),
        precision_score(y_test, y_pred_knn_pca),
        precision_score(y_test, y_pred_tree_original),
        precision_score(y_test, y_pred_tree_pca)
    ],
    'Recall': [
        recall_score(y_test, y_pred_logistic_original),
        recall_score(y_test, y_pred_logistic_pca),
        recall_score(y_test, y_pred_knn_original),
        recall_score(y_test, y_pred_knn_pca),
        recall_score(y_test, y_pred_tree_original),
        recall_score(y_test, y_pred_tree_pca)
    ],
    'F1 Score': [
        f1_score(y_test, y_pred_logistic_original),
        f1_score(y_test, y_pred_logistic_pca),
        f1_score(y_test, y_pred_knn_original),
        f1_score(y_test, y_pred_knn_pca),
        f1_score(y_test, y_pred_tree_original),
        f1_score(y_test, y_pred_tree_pca)
    ]
})

# Round the results
results[['Accuracy', 'Precision', 'Recall', 'F1 Score']] = results[
    ['Accuracy', 'Precision', 'Recall', 'F1 Score']
].round(4)

display(results)

# Part F: Hyperparameter Tuning
# Tune the Decision Tree using GridSearchCV

from sklearn.model_selection import GridSearchCV

# Define the parameter combinations to test
param_grid = {
    'max_depth': [3, 5, 7, 10, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'criterion': ['gini', 'entropy']
}

# Create a Decision Tree model
tree_for_tuning = DecisionTreeClassifier(random_state=42)

# Create GridSearchCV
grid_search = GridSearchCV(
    estimator=tree_for_tuning,
    param_grid=param_grid,
    cv=5,
    scoring='f1',
    n_jobs=-1
)

# Perform the search
grid_search.fit(X_train_scaled, y_train)

print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest Cross-Validation F1 Score:")
print(round(grid_search.best_score_, 4))

# Evaluate the tuned Decision Tree

best_tree = grid_search.best_estimator_

# Make predictions on the test data
y_pred_tree_tuned = best_tree.predict(X_test_scaled)

# Calculate performance metrics
tuned_accuracy = accuracy_score(y_test, y_pred_tree_tuned)
tuned_precision = precision_score(y_test, y_pred_tree_tuned)
tuned_recall = recall_score(y_test, y_pred_tree_tuned)
tuned_f1 = f1_score(y_test, y_pred_tree_tuned)

print("Tuned Decision Tree")
print("-------------------")
print("Accuracy :", round(tuned_accuracy, 4))
print("Precision:", round(tuned_precision, 4))
print("Recall   :", round(tuned_recall, 4))
print("F1 Score :", round(tuned_f1, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_tree_tuned))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_tree_tuned))

# Compare Decision Tree performance before and after tuning

tuning_comparison = pd.DataFrame({
    'Metric': [
        'Accuracy',
        'Precision',
        'Recall',
        'F1 Score'
    ],
    'Before Tuning': [
        accuracy_score(y_test, y_pred_tree_original),
        precision_score(y_test, y_pred_tree_original),
        recall_score(y_test, y_pred_tree_original),
        f1_score(y_test, y_pred_tree_original)
    ],
    'After Tuning': [
        tuned_accuracy,
        tuned_precision,
        tuned_recall,
        tuned_f1
    ]
})

display(tuning_comparison.round(4))

# Interpretation:
# Hyperparameter tuning improved accuracy, precision and F1 score.
# Recall decreased slightly after tuning.
# The increase in F1 score suggests that the tuned model provides
# a better overall balance between precision and recall.
# The selected parameters also restrict the complexity of the tree,
# which can help reduce overfitting and improve generalisation.

```
